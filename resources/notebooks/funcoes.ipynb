```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = (
    SparkSession.builder
    .appName("funcoes")
    .master("local[*]")
    .getOrCreate()
)
```

```python
schema = "Produtos STRING, Vendas INT"
vendas = [["caderno", 10],["Lápis", 20],["caderno", 5],["Caneta", 40], ["Mouse", 15], ["Lápis", 30], ["Teclado", 25]]
df = spark.createDataFrame(vendas, schema)
df.show()
```

```python
(df
 .groupBy("Produtos")
 .agg(F.sum("Vendas").alias("Total_Vendas"))
).show()
```

```python
df.show()
```

```python
df.select('Produtos', 'Vendas',
          F.expr('Vendas * 0.1 AS Comissao'),
          F.expr('Vendas * 2').alias('Vendas_Dobradas'),
          F.expr('Vendas * 0.1').alias('Desconto'),
          F.expr('Vendas - (Vendas * 0.1)').alias('Preco_Final')
          ).show()
```

```python
# variavel
valor_desconto = 10
vl_pct = valor_desconto / 100
print(valor_desconto, vl_pct)
```

```python
df.select('Produtos', 'Vendas',
          F.expr(f'Vendas * {vl_pct} AS Comissao'),
          F.expr(f'Vendas * {vl_pct}').alias('Desconto'),
          F.expr(f'Vendas - (Vendas * {vl_pct})').alias('Preco_Final')
          ).show()
```
