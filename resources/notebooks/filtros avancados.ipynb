```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = (
    SparkSession.builder
    .appName("Filtros")
    .master("local[*]")
    .getOrCreate()
)
```

```python
# select + where 

'''
.Select para selecionar colunas
== para igualdade
& para and, 
| para or, 
~ para not ,'
!=  para diferente, 
> para maior,
< para menor, 
>= para maior ou igual,
<= para menor ou igual
'''
```

```python
df_clientes = (
    spark.read.format("parquet")
    .load("/opt/spark/storage/output/customers.parquet")
)

df_clientes.show()
```

```python
clientes_colunas = (
    df_clientes.select("CustomerID", "Age", "City", "CustomerSegment")
)
clientes_colunas.show()
```

```python
"""
== para igualdade
& para and
"""

(
    df_clientes
    .select("CustomerID", "Age", "City", "CustomerSegment")
    .where (F.col("Age") == 65)
).show()
```

```python
"""
== para igualdade
& para and, 
"""

(
    df_clientes
    .select("CustomerID", "Age", "City", "CustomerSegment")
    .where ((F.col("Age") == 65) & (F.col("City") == "Mashhad"))
).show()
```

```python
"""
| para or, 
~ para not ,'
"""

(
    df_clientes
    .select("CustomerID", "Age", "City", "CustomerSegment")
    .where ((F.col("Age") == 18) | (F.col("CustomerID") == 100145))
).show()
```

```python
"""
!=  para diferente, 
> para maior,
< para menor, 
>= para maior ou igual,
<= para menor ou igual
"""

(
    df_clientes
    .select("CustomerID", "Age", "City", "CustomerSegment")
    .where ((F.col("Age") >= 18) & 
            ((F.col("Age") <= 24)))
).show()
```

```python
# teste de performance com filtros encadeados
teste1 = (
    df_clientes
    .select("CustomerID", "Age", "City", "CustomerSegment")
    .where (((F.col("Age") >= 18) & (F.col("Age") <= 24)) |
            ((F.col("Age") >= 60) & (F.col("Age") <= 65)))
)
teste1.explain(True)
```

```python
df2_clientes = spark.read.format("parquet").load("/opt/spark/storage/output/customers.parquet")
df2_clientes.show()
```

```python
df2_clientes.filter(F.col("Age").isin([18, 19, 20, 21, 22, 23, 24])).show() 
# tudo que estiver dentro da lista vai ser exibido
```

```python
idades_filter = [18, 19, 20, 21, 22, 23, 24]
df2_clientes.filter(F.col("Age").isin(idades_filter)).show()
```

```python
clientes_filter = [100001, 100002, 100003]
df2_clientes.filter(F.col("CustomerID").isin(clientes_filter)).show()
```

```python
# negação de filtro com ~

clientes_filter = [100001, 100002, 100003]
df2_clientes.filter(~F.col("CustomerID").isin(clientes_filter)).show()
```

```python
df2_clientes.filter(F.col("Age").isNull()).show()
```

```python
df2_clientes.filter(F.col("Age").isNull()).count()
```

```python
df2_clientes.groupBy("Age") \
    .count() \
    .orderBy("count", ascending=False) \
    .show()
```

```python
idade_mais_frequente = (df2_clientes
    .groupBy("Age")                         # Agrupa pela idade
    .count()                                # Conta quantas vezes cada idade aparece
    .orderBy("count", ascending=False)      # Ordena da mais frequente para a menos frequente
    .first()["Age"]                         # Pega a idade mais frequente
)

df2_clientes = df2_clientes.fillna({"Age": idade_mais_frequente})
```

```python
df2_clientes.filter(F.col("Age").isNull()).count()
```
