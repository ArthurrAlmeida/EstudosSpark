```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = (
    SparkSession.builder
    .appName("Datas")
    .master("local[*]")
    .config("spark.driver.memory", "2g")
    .config("spark.executor.memory", "2g")
    .getOrCreate()
)
spark.conf.set("spark.sql.session.timeZone", "America/Sao_Paulo")
```

```python
df = spark.read.format("csv").option("header", "true").load("/opt/spark/storage/E-Commerce/orders.csv")
df.show()
```

```python
df.printSchema()
```

```python
df = df.withColumn("convertendo_timestamp", F.to_timestamp("OrderDate"))
```

```python
df.show()
df.printSchema()
```

```python
df = df.withColumn("convertendo_date", F.to_date("OrderDate"))
```

```python
df.show()
df.printSchema()
```

```python
# usando data e hora atual do sistema operacional

(
    df
    .withColumn("data_hora_atual", F.current_timestamp())
    .withColumn("data_atual", F.current_date())
).show(5)
```

```python
(
    df
    .withColumn("data_hora_atual", F.current_timestamp())
    .withColumn("data_atual", F.current_date())
    .withColumn("diferenca_dias", F.datediff("data_atual", "OrderDate"))
    .withColumn("diferenca_meses", F.floor(F.months_between("data_atual", "OrderDate")))
    .withColumn("diferenca_meses_arredondando_cima", F.ceil(F.months_between("data_atual", "OrderDate")))
    .withColumn("diferenca_meses_arredondando_mat", F.round(F.months_between("data_atual", "OrderDate")))
).show(5)
```

```python
# exemplos

df_testes =(df
    .withColumn("data_hora_atual", F.current_timestamp())
    .withColumn("data_br_completo", F.date_format("data_hora_atual", "dd/MM/yyyy HH:mm:ss"))
    .withColumn(
        "prazo_entrega",
        F.date_format(F.date_add("data_hora_atual", 5), "dd/MM/yyyy")
    )
)
df_testes.show(5)
```

```python
df_testes2 = (
    df_testes.select(
            "OrderID", "CustomerID", "ProductID", 
            "Quantity", "Discount", "PaymentMethod", 
            "Status", "data_br_completo", "prazo_entrega",
            F.date_add("data_hora_atual", 7).alias("limite_final_entrega"),
            F.dayofmonth("limite_final_entrega").alias("dia_mes_atual")
            )
)

df_testes2.show(5)
```
