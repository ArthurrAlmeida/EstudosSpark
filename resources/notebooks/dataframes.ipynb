```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = (
    SparkSession.builder
    .appName("Datagrames")
    .master("local[*]")
    .getOrCreate()
)
```

```python
df1 = spark.createDataFrame([("Arthur", 25), ("Maria", 30), ("João", 22)])
df1.show()
```

```python
schema = "Id INT, Nome STRING"
dados = [(1, "Arthur"), (2, "Maria"), (3, "João")]
df2 = spark.createDataFrame(dados, schema)
df2.show()
```

```python
df2.printSchema()
```
